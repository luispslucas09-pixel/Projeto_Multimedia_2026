```python
# Pergunta inicial para saber se o utilizador quer jogar
jogar = input('Vamos Jogar? s/n: ')
score = 0

# Lista com todas as perguntas, respetiva resposta correta e texto de correção
perguntas = [
    {"texto": "Pergunta 1 - Qual a Capital de Portugal? ", "resp": "lisboa", "corrigida": "Lisboa"},
    {"texto": "Pergunta 2 - Quantos oceanos existem no planeta? ", "resp": "5", "corrigida": "5"},
    {"texto": "Pergunta 3 - Qual a capital da Holanda? ", "resp": "amesterdão", "corrigida": "Amesterdão"},
    {"texto": "Pergunta 4 - Qual a capital da Grécia? ", "resp": "atenas", "corrigida": "Atenas"},
    {"texto": "Pergunta 5 - Qual é o país de capital Berna? ", "resp": "suíça", "corrigida": "Suíça"},
    {"texto": "Pergunta 6 - Qual é o maior continente do mundo em área? ", "resp": "ásia", "corrigida": "Ásia"},
    {"texto": "Pergunta 7 - Qual é o rio mais longo do mundo? ", "resp": "rio nilo", "corrigida": "Rio Nilo"},
    {"texto": "Pergunta 8 - Qual é o ponto mais alto do planeta Terra? ", "resp": "monte everest", "corrigida": "Monte Everest"},
    {"texto": "Pergunta 9 - Qual o menor país do mundo em extensão territorial? ", "resp": "vaticano", "corrigida": "Vaticano"},
    {"texto": "Pergunta 10 - Qual é a linha imaginária que divide a Terra em Hemisfério Norte e Hemisfério Sul? ", "resp": "linha do equador", "corrigida": "Linha do Equador"}
]

total_de_perguntas = len(perguntas) # Calcula automaticamente o total de perguntas

if jogar.lower() == 's':
    # Ciclo 'for' que percorre a lista de perguntas uma a uma
    for p in perguntas:
        resposta = input(p["texto"]).strip().lower()
        
        if resposta == p["resp"]:
            score += 1
            print('Certo!')
        else:
            print(f'Resposta Errada, é {p["corrigida"]}')

    # Apresentação dos resultados finais
    print(f'\nObrigado por jogares este QUIZ! Obtiveste {score} Resposta(s) Certa(s)')
    
    percentagem = (score / total_de_perguntas) * 100
    print(f'A tua percentagem é de {percentagem}%')

print('Jogo Terminado')
```

```python
# Inicialização da lista de tarefas em memória
tarefas = ["Terminar o README.md"]
print("Gestor de Tarefas inicializado com sucesso!")

def mostrar_tarefas():
    if not tarefas:
        print("\nA tua lista de tarefas está vazia!")
    else:
        print("\nLista de Tarefas a Fazer:")
        for i, tarefa in enumerate(tarefas):
            print(f"  {i + 1}. [ ] {tarefa}")

mostrar_tarefas()

# Podes executar esta célula sempre que quiseres adicionar uma nova tarefa
nova_tarefa = input("Escreve a nova tarefa a adicionar: ")

if nova_tarefa.strip():
    tarefas.append(nova_tarefa)
    print(f'Tarefa "{nova_tarefa}" adicionada com sucesso!')
else:
    print("A tarefa não pode estar vazia.")

mostrar_tarefas()

# Executa esta célula para concluir/remover uma tarefa pelo seu número
mostrar_tarefas()

try:
    escolha = input("\nIntroduz o número da tarefa que concluíste (ou prime Enter para sair): ")
    if escolha.strip():
        indice = int(escolha) - 1
        if 0 <= indice < len(tarefas):
            tarefa_removida = tarefas.pop(indice)
            print(f'Parabéns! Tarefa "{tarefa_removida}" concluída e removida.')
            mostrar_tarefas()
        else:
            print("Número inválido.")
    else:
        print("Nenhuma tarefa foi alterada.")
except ValueError:
    print(" Por favor, introduz um número válido.")
```

```python
# ==========================================
# FUNÇÃO: MOSTRAR TAREFAS
# Legenda: Esta função serve para exibir a lista atual de tarefas no ecrã.
# ==========================================
def mostrar_tarefas(tarefas):
    # Verifica se a lista está vazia
    if not tarefas:
        print("\n📋 A tua lista de tarefas está vazia!")
    else:
        print("\n📋 Lista de Tarefas a Fazer:")
        # O 'enumerate' numera cada item da lista começando em 1
        for i, tarefa in enumerate(tarefas, 1):
            print(f"  {i}. [ ] {tarefa}")

# ==========================================
# FUNÇÃO PRINCIPAL DO PROGRAMA
# ==========================================
def main():
    # 1. Inicialização da lista de tarefas em memória
    tarefas = ["Terminar o README.md"]
    print("🚀 Gestor de Tarefas inicializado com sucesso!")
    
    # 2. Ciclo 'while True' mantém o menu a funcionar repetidamente até o utilizador escolher sair
    while True:
        print("\n====================")
        print("    GESTOR DE TAREFAS")
        print("====================")
        print("1. Ver tarefas")
        print("2. Adicionar tarefa")
        print("3. Concluir / Remover tarefa")
        print("4. Sair")
        
        # Lê a opção escolhida pelo utilizador e remove espaços extras com .strip()
        opcao = input("\nEscolhe uma opção (1-4): ").strip()
        
        # Opção 1: Mostrar a lista de tarefas
        if opcao == '1':
            mostrar_tarefas(tarefas)
            
        # Opção 2: Adicionar uma nova tarefa à lista
        elif opcao == '2':
            nova_tarefa = input("Escreve a nova tarefa a adicionar: ").strip()
            # Valida se o utilizador escreveu mesmo alguma coisa
            if nova_tarefa:
                tarefas.append(nova_tarefa)  # Adiciona ao final da lista
                print(f'✅ Tarefa "{nova_tarefa}" adicionada com sucesso!')
            else:
                print("⚠️ A tarefa não pode estar vazia.")
                
        # Opção 3: Concluir ou remover uma tarefa
        elif opcao == '3':
            mostrar_tarefas(tarefas)
            if tarefas:
                try:
                    escolha = input("\nIntroduz o número da tarefa que concluíste (ou prime Enter para voltar): ").strip()
                    if escolha:
                        # Converte o número digitado para índice de lista (subtrai 1)
                        indice = int(escolha) - 1
                        # Valida se o número está dentro dos limites da lista
                        if 0 <= indice < len(tarefas):
                            tarefa_removida = tarefas.pop(indice)  # Remove a tarefa pelo índice
                            print(f'🎉 Parabéns! Tarefa "{tarefa_removida}" concluída e removida.')
                        else:
                            print("❌ Número inválido.")
                except ValueError:
                    # Captura o erro caso o utilizador digite letras em vez de números
                    print("❌ Por favor, introduz um número válido.")
                    
        # Opção 4: Sair do programa
        elif opcao == '4':
            print("👋 Obrigado por usares o Gestor de Tarefas. Até logo!")
            break  # Interrompe o ciclo 'while' e termina o programa
            
        else:
            print("❌ Opção inválida. Escolhe um número entre 1 e 4.")

# Ponto de entrada do programa em Python
if __name__ == "__main__":
    main()
```

```python
def calcular_recursos_podcast(episodios_mes, duracao_media_min):
    minutos_totais = episodios_mes * duracao_media_min
    
    # Cálculos técnicos de áudio
    espaco_mb = minutos_totais * 1.44          # 1.44 MB/min (MP3 192kbps)
    horas_edicao = (minutos_totais * 2.5) / 60 # Fator de edição 2.5x
    custo_estimado = (episodios_mes * 15) + (minutos_totais * 0.50)
    
    return espaco_mb, horas_edicao, custo_estimado

def gerar_guiao_relampago(tema):
    return f"""
    📄 GUIÃO RELÂMPAGO: {tema.upper()}
    --------------------------------------------------
    1. INTRODUÇÃO: Olá! Hoje vamos explorar tudo sobre '{tema}' e como isso muda o teu dia a dia.
    2. TÓPICOS PRINCIPAIS:
       - Tópico 1: O contexto atual de {tema}.
       - Tópico 2: Os erros mais comuns a evitar.
       - Tópico 3: Estratégias práticas para aplicar já hoje.
    3. CHAMADA PARA AÇÃO: Subscreve o canal e partilha este episódio com alguém que precise de ouvir isto!
    """

# Exemplo de execução
ep = int(input("Quantos episódios vais gravar por mês? "))
dur = int(input("Qual a duração média de cada episódio (minutos)? "))
ideia = input("Qual a ideia central do episódio? ")

espaco, edicao, custo = calcular_recursos_podcast(ep, dur)

print(f"\n📊 RESULTADOS MENSAIS:")
print(f"💾 Espaço necessário: {espaco:.1f} MB")
print(f"✂️ Tempo estimado de edição: {edicao:.1f} horas")
print(f"💶 Orçamento estimado: {custo:.2f} €")
print(gerar_guiao_relampago(ideia))
```
